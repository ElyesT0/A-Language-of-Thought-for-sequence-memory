```python
# --------------------
# READ PROCESSED FILE
# --------------------
# -- Retrieving the pre-processed Data
data_main=pd.read_csv(path_processed_data, index_col=False)

# ***
# -- Formatting
# ***

# Drop the training rows
data_main=data_main.drop(data_main[data_main['seq_name'] == 'Training'].index)
data_main.reset_index(inplace=True, drop=True)

# Turn the arrays into numbers
data_main=str2int_dataset(data_main,processed=True, label_int_col=label_int_col)

# Dataframe with only the participants who completed the full experiment/or that have the same amount of trials as max
data_main_complete=data_main.copy()
```

```python
population_size=len(np.unique(data_main_complete['participant_ID']))

# To get the durations of the experiment, we get the values of last entry timepoints (in ms)
last_trial=[]
# -- We need an array that will hold the IDs of the participants to keep track of which timepoints we need to take
IDs=[data_main_complete["participant_ID"][0]]
for i in range(len(data_main_complete)-1):
    if data_main_complete.iloc[i]["participant_ID"] not in IDs:
        IDs.append(data_main_complete.iloc[i]["participant_ID"])
        last_trial.append(data_main_complete.iloc[i-1]["click_timings_after"][-1])
last_trial.append(data_main_complete.iloc[len(data_main_complete)-1]["click_timings_after"][-1])
readable_last_trial=[i/60000 for i in last_trial]
print("Mean duration of the experiment is: {}".format(np.mean(readable_last_trial)))
print("Median duration of the experiment is: {}".format(np.median(readable_last_trial)))
print("Maximum duration of the experiment is: {}".format(np.max(readable_last_trial)))
print("Number of participants is: {}".format(population_size))
print('*******************************************\n')


```

# Experiment 2

```python
# --------------------
# READ PROCESSED FILE
# --------------------
# -- Retrieving the pre-processed Data
data_main_exp2=pd.read_csv(path_processed_data_exp2, index_col=False)

# ***
# -- Formatting
# ***

# Drop the training rows
data_main_exp2=data_main_exp2.drop(data_main_exp2[data_main_exp2['seq_name'] == 'Training'].index)
data_main_exp2.reset_index(inplace=True, drop=True)

# Turn the arrays into numbers
data_main_exp2=str2int_dataset(data_main_exp2,processed=True, label_int_col=label_int_col)

# --- Dataframe with only the participants who completed the full experiment/or that have the same amount of trials as max
# Get the original number of rows
original_row_count = len(data_main_complete_exp2)

# Count the number of rows per unique 'participant_ID'
participant_counts = data_main_complete_exp2['participant_ID'].value_counts()

# Get the index of participants with exactly 50 rows
participants_with_50_rows = participant_counts[participant_counts == 50].index

# Filter data_main_complete_exp2 in place and reset the index
data_main_complete_exp2 = data_main_complete_exp2[data_main_complete_exp2['participant_ID'].isin(participants_with_50_rows)].reset_index(drop=True)

# Calculate and print the number of dropped rows
dropped_row_count = original_row_count - len(data_main_complete_exp2)
print("Number of dropped rows:", dropped_row_count)

```

```python
population_size=len(np.unique(data_main_complete_exp2['participant_ID']))

# To get the durations of the experiment, we get the values of last entry timepoints (in ms)
last_trial=[]
# -- We need an array that will hold the IDs of the participants to keep track of which timepoints we need to take
IDs=[data_main_complete_exp2["participant_ID"][0]]
for i in range(len(data_main_complete_exp2)-1):
    if data_main_complete_exp2.iloc[i]["participant_ID"] not in IDs:
        IDs.append(data_main_complete_exp2.iloc[i]["participant_ID"])
        last_trial.append(data_main_complete_exp2.iloc[i-1]["click_timings_after"][-1])
last_trial.append(data_main_complete_exp2.iloc[len(data_main_complete_exp2)-1]["click_timings_after"][-1])
readable_last_trial=[i/60000 for i in last_trial]
print("Mean duration of the experiment is: {}".format(np.mean(readable_last_trial)))
print("Median duration of the experiment is: {}".format(np.median(readable_last_trial)))
print("Maximum duration of the experiment is: {}".format(np.max(readable_last_trial)))
print("Number of participants is: {}".format(population_size))
print('*******************************************\n')

# Finding if there is any overlap in the data
id_exp1=np.array(data_main_complete['participant_ID'].unique())
id_exp2=np.array(data_main_complete_exp2['participant_ID'].unique())
common_values=np.intersect1d(id_exp1,id_exp2)
print(f'There are {len(common_values)} participant id that are both in exp1 and exp2')
```
