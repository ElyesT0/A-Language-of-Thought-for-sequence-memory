```python
import numpy as np
import pandas as pd
from modules import *
import scipy.stats as stats
import os

pd.set_option('display.max_rows', 100)
pd.set_option('display.max_columns', None)

%load_ext autoreload
%autoreload 2
```

```python
data_filename='experiment1_processed_10h05_31032023_data.csv'
root_path='/Users/et/Documents/UNICOG/2-Experiments/memocrush'
path_processed_data=os.path.join(root_path,f'Data/processed/experiment-1/{data_filename}')
plot_path=os.path.join(root_path,'Figures')
```

```python
# --------------------
# READ PROCESSED FILE
# --------------------
# -- Retrieving the pre-processed Data
data_main=pd.read_csv(path_processed_data, index_col=False)

# ***
# -- Formatting
# ***

# Drop the training rows
data_main=data_main.drop(data_main[data_main['seq_name'] == 'Training'].index)
data_main.reset_index(inplace=True, drop=True)

# Turn the arrays into numbers
data_main=str2int_dataset(data_main,processed=True, label_int_col=label_int_col)

# Dataframe with only the participants who completed the full experiment/or that have the same amount of trials as max
data_main_complete=data_main.copy()
```

```python
population_size=len(np.unique(data_main_complete['participant_ID']))

# To get the durations of the experiment, we get the values of last entry timepoints (in ms)
last_trial=[]
# -- We need an array that will hold the IDs of the participants to keep track of which timepoints we need to take
IDs=[data_main_complete["participant_ID"][0]]
for i in range(len(data_main_complete)-1):
    if data_main_complete.iloc[i]["participant_ID"] not in IDs:
        IDs.append(data_main_complete.iloc[i]["participant_ID"])
        last_trial.append(data_main_complete.iloc[i-1]["click_timings_after"][-1])
last_trial.append(data_main_complete.iloc[len(data_main_complete)-1]["click_timings_after"][-1])
readable_last_trial=[i/60000 for i in last_trial]
print("Mean duration of the experiment is: {}".format(np.mean(readable_last_trial)))
print("Median duration of the experiment is: {}".format(np.median(readable_last_trial)))
print("Maximum duration of the experiment is: {}".format(np.max(readable_last_trial)))
print("Number of participants is: {}".format(population_size))
print('*******************************************\n')


```
