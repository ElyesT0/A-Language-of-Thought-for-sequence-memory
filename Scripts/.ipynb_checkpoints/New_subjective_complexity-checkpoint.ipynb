<h1 style = 'background-color: #800020; color: white; padding:10px;'>
Clean Dataset
</h1>

```python
# Go over probe-easy trials
mask_easy = "sequences_temp_tags == 'probe-easy'"
df_filtered = subj_dataset.query(mask_easy)
dict_flagged = {}
flagged = []
arr_response = []
holder = []
previous_p = 0

for _, row in df_filtered.iterrows():
    response = row['participant_response']
    p = row['participant_prolific_id']

    if response > 2:
        if previous_p and previous_p != p:  # New participant encountered
            arr_response.append(holder)
            dict_flagged[previous_p] = holder
            holder = []
        flagged.append(p)
        holder.append(response)
        previous_p = p

# Flush the last participant's data after the loop
if holder:
    arr_response.append(holder)
    dict_flagged[previous_p] = holder

flagged = np.unique(flagged)
excluded = []
for key,val in dict_flagged.items():
    if len(val)>2:
        excluded.append(key)

subj_data_clean = subj_dataset.drop(subj_dataset[subj_dataset['participant_prolific_id'].isin(excluded)].index)
all_sequences = subj_data_clean['sequences_temp_tags'].unique()
n_sequences = len(all_sequences)

valid_participants = (
    subj_data_clean.groupby('participant_prolific_id')['sequences_temp_tags']
    .nunique()
    .pipe(lambda s: s[s == n_sequences].index)
)

subj_data_clean = subj_data_clean[subj_data_clean['participant_prolific_id'].isin(valid_participants)]
all_durations = (subj_data_clean.groupby('participant_prolific_id')
                 .apply(lambda g: (g.iloc[0].last_click - g.iloc[0].participant_startTime)/(60000))
                 .tolist()
                )
mean_duration = np.mean(all_durations)
print('Initial N : ', len(subj_dataset.participant_prolific_id.unique()))
print("excluded : ",len(subj_dataset.participant_prolific_id.unique()) - len(subj_data_clean.participant_prolific_id.unique()))
print('Final dataset contains : ', len(subj_data_clean.participant_prolific_id.unique()), ' participants')
print('Mean Duration : ', mean_duration.round(2), 'mins')
subj_data_clean_with_probes = subj_data_clean.copy()
mask = subj_data_clean.sequences_temp_tags.str.contains('probe-easy|probe-hard|training')
subj_data_clean = subj_data_clean[~mask].reset_index(drop=True)



# -------------------------
seq_name_mapping = {
    'Rep2':           'Repetition-2',
    'Rep3':           'Repetition-3',
    'Rep4':           'Repetition-4',
    'Rep-Nested':     'Repetition-Nested',
    'Mirror-Rep':     'Mirror-Rep',
    'Mirror-NoRep':   'Mirror-NoRep',
    'Insertion':      'Insertion',
    'Suppression':    'Suppression',
    'Play':           'play',
    'Play4':          'play 4 tokens',
    'Index':          'index i',
    'Sub-1':          'sub-programs 1',
    'Sub-2':          'sub-programs 2',
    'CRep2':          'control Repetition-2',
    'CRep3':          'control Repetition-3',
    'CRep4':          'control Repetition-4',
    'CMirror-Rep':    'control Mirror-Rep',
    'CMirror-NoRep':  'control Mirror-NoRep',
    'CPlay':          'control play',
    'CPlay4':         'control play 4 tokens',
    'CIndex':         'control index i',
    'CSub-1':         'control sub-programs 1',
    'CSub-2':         'control sub-programs 2',
    'Nested-Global-Rep' : 'control NoLocal nested',
    'Nested-Local-Rep' : 'control NoGlobal nested',
}
```

Analyze

<h1 style='background-color: #800020; color: white; padding : 10px;'>
Analysis
</h1>

```python
# Mean per participant per sequence
per_participant_mean = (
    subj_data_clean
    .groupby(['sequences_temp_tags', 'participant_prolific_id'])['participant_response']
    .mean()
)

# Then mean of means + SEM across participants, per sequence
summary = (
    per_participant_mean
    .groupby('sequences_temp_tags')
    .agg(
        mean='mean',
        sem=lambda x: x.sem(),
        N = 'count'
    ).round({'mean' : 2, 'sem' :2})
)
summary.index = summary.index.map(seq_name_mapping)

df_results_exp2['subjective_complexity'] = df_results_exp2['seq_name'].map(summary['mean'])
```
