```python
def individual_interclick_and_accuracy(data,path,name_index,start_index=0, x_axis_num=False, ylim_min=0.3, ylim_max=1.4):
    # Holds all the responses' interclicks that are correct until item of index "step"
    holder_subsequence_interclicks=[]
    
    # Holds the last mean interclick timings of the holder_subsequence_interclicks. Will be used for plotting.
    holder_mean_interclicks=[]
    
    # Holds the Standard error of the mean for each mean timing.
    holder_sem_interclicks=[]
    
    # Select relevant data
    # -- 1. Subset based on sequence name
    subset_name=data[data['seq_name']==seq_name_list[name_index]].copy()
    

    accuracy = []
    for participant in data['participant_ID'].unique():
        subset_participant = data[data['participant_ID'] == participant]
        total_nb_responses = len(subset_participant)
        participant_accuracy = []
        for i in range(1, max_positions + 1):
            correct_count = 0
            valid_responses = 0
            for index, row in subset_participant.iterrows():
                response = row['response_structure'] if structure else row['sequences_response']
                origin=row['sequences_structure'] if structure else row['seq']
                if len(origin) >= i and len(response) >= i:
                    valid_responses += 1
                    if origin[:i] == response[:i]:
                        correct_count += 1
            participant_accuracy.append(correct_count / total_nb_responses if total_nb_responses > 0 else 0)
        accuracy.append(participant_accuracy)

     # Calculate and plot accuracy histogram
    accuracy = compute_accuracy(subset_name)
    mean_accuracy = np.mean(accuracy, axis=0)
    sem_accuracy = np.std(accuracy, axis=0) / np.sqrt(len(accuracy))

    # Initialize my plotting objects
    all_cumulative_interclicks=[]
    all_sem_interclick=[]
    all_accuracies=[]
    all_sem_accuracies=[]

    for step in range(1,12):
        if step<=start_index:
        # -- 2. Subset based on correct answers up to index number 'step'
            subset_correct = subset_name[
            (subset_name['sequences_structure'].apply(len) > start_index) & 
            (subset_name['comparable_temp'].apply(len) > start_index) & 
            (subset_name['sequences_structure'].apply(lambda x: x[:start_index]) == subset_name['comparable_temp'].apply(lambda x: x[:start_index]))
        ] 
        else:
            # -- 2. Subset based on correct answers up to index number 'step'
            subset_correct = subset_name[
            (subset_name['sequences_structure'].apply(len) > step) & 
            (subset_name['comparable_temp'].apply(len) > step) & 
            (subset_name['sequences_structure'].apply(lambda x: x[:step]) == subset_name['comparable_temp'].apply(lambda x: x[:step]))
        ] 

                
        if len(subset_correct)==0:
            holder_subsequence_interclicks=0
            holder_mean_interclicks.append(np.nan)
            holder_sem_interclicks.append(np.nan)   

            
            
        else:
                
            # Fill the holders
            holder_subsequence_interclicks = subset_correct['interclick_time'].apply(lambda x: np.array(x[:step]) / 1000).to_numpy()  # Convert to seconds
            last_interclicks_column=[arr[-1] for arr in holder_subsequence_interclicks]
            mean_temp=np.mean(last_interclicks_column)
            holder_mean_interclicks.append(mean_temp)
            holder_sem_interclicks.append(np.std(last_interclicks_column)/np.sqrt(len(last_interclicks_column)))
            
                
    # Append results to plotting objects
    all_cumulative_interclicks=np.array(holder_mean_interclicks)
    all_sem_interclick=np.array(holder_sem_interclicks)
    all_accuracies=np.array(mean_accuracy)
    all_sem_accuracies=np.array(sem_accuracy)
    
    # Plotting interclicks
    #
    # ---------- Plot interclicks -------------
    #
    plt.figure(figsize=(10, 6))

    # Create a single set of axes
    ax = plt.gca()
    
    # Plot vertical lines
    ax.vlines(x=range(11), ymin=ylim_min, ymax=ylim_max, colors='black', ls='--', lw=1)
    
    # Define the labels used in the x-axis. Either letters constitutive of the sequence structure or simple indexes.
    if x_axis_num:
        ax.set_xticks(ticks=range(11), labels=range(1, 12))
    else:
        ax.set_xticks(ticks=[i - 0.5 for i in range(12)], labels=[i for i in alpha_seq_expression[name_index]])
        ax.set_xlim(xmin=-1, xmax=11)
    
    ax.set_title(f'{seq_name_list[name_index]}', fontweight='bold')
    ax.errorbar(range(11), all_cumulative_interclicks, yerr=all_sem_interclick, fmt='o', capsize=5, capthick=2, color="black")
    ax.plot(range(11), all_cumulative_interclicks)
    ax.set_ylim(ymin=ylim_min, ymax=ylim_max)  # Set y-axis limits
    
    # Show y-tick labels only if needed
    ax.set_ylabel("Interclick \nduration (s)", rotation=0, fontstyle='italic')
    ax.yaxis.set_label_coords(-0.2, 0.92)  # Position the label at the top
    
    # ---------- Plot Accuracies -------------
    # Add secondary y-axis for histogram
    ax2 = ax.twinx()
    
    ax2.bar(
        [i - 0.5 for i in range(0, 12)],  # X positions
        all_accuracies,  # Heights
        yerr=all_sem_accuracies,  # Error bars
        capsize=5,  # Size of the caps on the error bars
        alpha=0.3,  # Transparency of the bars themselves
        color='gray',  # Color of the bars
        error_kw=dict(elinewidth=2, alpha=0.5)  # Customizing error bars: line width and transparency
    )
    ax2.set_ylim(0, 1)
    
    # Set the secondary y-axis label
    ax2.set_ylabel("Accuracy on \nn first items", rotation=0, fontstyle='italic')
    ax2.yaxis.set_label_coords(1.25, 1.02)  # Position the label at the top
    
    # Save and display the plot
    plt.savefig(f'{path}/interclick/cumulated_accuracy/{name_index}_{seq_name_list[name_index]}_mean_interclicks_with_accuracy_start-{start_index}.png', bbox_inches='tight', dpi=800)
    plt.show()

    
```

## Modular vs Continous memorization

It could be that (2.A.) chunks are read with rules application or individual vocabulary elements are read with a statistical learning system. (2.B.) It is also possible that the reading occurs several times serially at different stages (e.g., first on vocabulary items, then on chunks that are the new mental elements). ⇒ Look at recall accuracy for [:n] step by step to see if it’s modular, i.e. we recall by groups of items, or continuous, we recall item by item. Continous effect are primacy effects for example, but if we observe a sudden drop in accuracy we can confidently say that the chunk was a mental item.

We can decompose the representation of subprogram-2 <br>
I believe there are two encoding path: LoT and 'No rule found'<br><br>
<b>If we decompose the levels of the 'No rule found' system:</b> <ol>
<li>Cannot find two occurences of "ABC" -> We remove responses that fit No-Rule_2 and No-Rule3 </li><br>
<li>We have {A,B,C,D,E,F} that are represented more or less equally. (ex: AEBDFCE) -> We remove responses that fit No-Rule_3<br>[(x)subprogram,(x)ordinal position,(v)vocabulary]</li><br>
<li>We have {A,B,C,D,E,F} represented equally and in the right order. Maybe participant perceived there are more A,B,Cs but it is not structured. (ex: ABBCDAEF) <br>[(x)subprogram,(v)ordinal position,(v)vocabulary]</li><br> </ol>

<br><b>If we decompose the levels of the 'LoT-Rule' system:</b> <ol>
<li>We find two or more repetitions of the ABC chunk. It does not alternate with items from secondary vocabulary {D,E,F} (ex: 012.32.012.2310) -> We remove responses that fit LoT-Rule_2 and LoT-Rule3 <br> [(x)subprogram, (v)repetition rule, (x)ordinal position,(x)vocabulary]</li><br>
<li>We find two or more repetitions of the ABC chunk. It alternates in a 3-1 manner with items from secondary vocabulary {D,E,F}. Not necessarily in the right order for secondary voc and maybe secondary voc is incompletely expressed (ex: 012.3.012.5.012.3) -> We remove responses that fit LoT-Rule3 <br> [(x)subprogram, (v)repetition rule, (x)ordinal position,(v)vocabulary]</li><br>
<li>Perfect reproduction.<br> [(v)subprogram, (v)repetition rule, (v)ordinal position,(v)vocabulary]<br></li>

question: in non-structured sequences, are the averaged reaction times longer than in structured sequences? If YES what could be a possible explanation ? => there is a larger range of possible answers in the vocabulary. <br>
Compare CONTROL sequences with local repetition of same element (interclick timing) and CONTROL sequences with no-local repetitions of the same element. Can we show that the participants, even when they did not grasp the structure, work with negative rules (no-local repetitions) by both seeing that there is a <ul><li>shorter interclick duration when no local repetition (suggesting that scanning the array of possible next element is shorter),</li><li>no local repetitions in responses when no local rep in original + local repetition on wrong items in controls (transfer of rule)</li></ul>

```python
print('----------------------------------------------------')
print('All structured vs All controls: interclick-duration')
print('----------------------------------------------------\n')
# Interclick duration in controls with no local repetition
#
mean_interclick_structured_allParticipants=[]
mean_interclick_control_allParticipants=[]

data=data_main_complete.copy()
subset_structured = data[~data['seq_name'].str.contains('control')][['seq_name','participant_ID','interclick_time','LoT Complexity','sequences_structure']]
subset_control=data[data['seq_name'].str.contains('control')][['seq_name','participant_ID','interclick_time','LoT Complexity','sequences_structure']]

# - 0 - Remove outliers (especially efficient to remove long pauses at the start or end of the sequence)
subset_structured_clean=subset_structured.copy()
subset_control_clean=subset_control.copy()
subset_structured_clean['interclick_time']=subset_structured['interclick_time'].apply(remove_outliers)
subset_control_clean['interclick_time']=subset_control['interclick_time'].apply(remove_outliers)

# - 1 - Mean Interclick duration for structured sequences vs control sequences

# __first per participant
# Structured
for participant in subset_structured_clean['participant_ID'].unique():
    subset_structured_participant=subset_structured_clean[subset_structured_clean['participant_ID']==participant]
    mean_interclick_structured_allParticipants.append(np.mean(np.concatenate(subset_structured_participant['interclick_time'].values)))

# Controls
for participant in subset_control_clean['participant_ID'].unique():
    subset_control_participant=subset_control_clean[subset_control_clean['participant_ID']==participant]
    mean_interclick_control_allParticipants.append(np.mean(np.concatenate(subset_control_participant['interclick_time'].values)))

# __Global mean
    
# Structured
mean_structured=np.mean(mean_interclick_structured_allParticipants)
std_structured=np.std(mean_interclick_structured_allParticipants)
sem_structured = std_structured / np.sqrt(len(mean_interclick_structured_allParticipants))

# Controls
mean_control=np.mean(mean_interclick_control_allParticipants)
std_control=np.std(mean_interclick_control_allParticipants)
sem_control = std_control / np.sqrt(len(mean_interclick_control_allParticipants))

# - 2 - Statistic test to compare both: 
#
# -- 2a. Test for Normality (Parametric/non-parametric test)
stat, p_value = kstest(mean_interclick_structured_allParticipants, 'norm')
print(f'Result of Kolmogorov-Smirnov test for normality on STRUCTURED sequences: {stat}, p-value: {p_value}')

stat, p_value = kstest(mean_interclick_control_allParticipants, 'norm')
print(f'Result of Kolmogorov-Smirnov test for normality on CONTROL sequences: {stat}, p-value: {p_value}')

# -- 2b. Those are dependent samples (so Wilcoxon signed-rank test if Non-parametric, Paired t-test if parametric)

# - 3 - Print results
stat, p_value = wilcoxon(mean_interclick_structured_allParticipants,mean_interclick_control_allParticipants)
print(f'\nWilcoxon stat: {stat}, p_value : {p_value}')
print(f'Mean Interclick duration. STRUCTURED: {round(mean_structured,3)}, CONTROL: {round(mean_control,3)}')
print(f'Standard Deviation. STRUCTURED: {round(std_structured,3)}, CONTROL: {round(std_control,3)}')
print(f'SEM. STRUCTURED: {round(sem_structured,3)}, CONTROL: {round(sem_control,3)}')
print('--\n--\n--\n')


print('--------------------------------------------------------------')
print('Control-MirrorRep vs Control sub-program-1: Local repetitions')
print('--------------------------------------------------------------\n')

# Get subset data
subset_subprog=data[data['seq_name']==seq_name_list[12]][['sequences_structure','comparable_temp','performance']]
subset_mirror=data[data['seq_name']==seq_name_list[22]][['sequences_structure','comparable_temp','performance']]
subset_alternate=data[data['seq_name']==seq_name_list[4]][['sequences_structure','comparable_temp','performance']]


# Count number of local repetitions in the responses while checking if we are not counting the pair twice
def count_identical_pairs(arr):
    counts = [0, 0, 0, 0]
    contains_pair = [False, False, False, False]
    i = 0
    while i < len(arr) - 1:
        if arr[i] == arr[i + 1]:
            counts[arr[i]] += 1
            contains_pair[arr[i]] = True
            i += 2  # Move to the next pair after counting this one
        else:
            i += 1  # Move to the next element
    return counts, contains_pair

# Apply the function to each array in the column
pair_counts_mirror = subset_mirror['comparable_temp'].apply(lambda x: count_identical_pairs(x)[0])
pair_counts_subprog = subset_subprog['comparable_temp'].apply(lambda x: count_identical_pairs(x)[0])
pair_counts_alternate = subset_alternate['comparable_temp'].apply(lambda x: count_identical_pairs(x)[0])

pair_bool_mirror = subset_mirror['comparable_temp'].apply(lambda x: count_identical_pairs(x)[1])
pair_bool_subprog = subset_subprog['comparable_temp'].apply(lambda x: count_identical_pairs(x)[1])
pair_bool_alternate = subset_alternate['comparable_temp'].apply(lambda x: count_identical_pairs(x)[1])

# Sum the results to get the total counts for each item
count_local_mirror = np.sum(pair_counts_mirror.tolist(), axis=0)
count_local_subprog = np.sum(pair_counts_subprog.tolist(), axis=0)
count_local_alternate = np.sum(pair_counts_alternate.tolist(), axis=0)

percentage_local_mirror = np.round(np.mean(pair_bool_mirror.tolist(), axis=0) * 100, 2)
percentage_local_subprog = np.round(np.mean(pair_bool_subprog.tolist(), axis=0) * 100, 2)
percentage_local_alternate = np.round(np.mean(pair_bool_alternate.tolist(), axis=0) * 100, 2)


# Print results
alphabet_list=['A','B','C','D']
print(f'**Counting Local repetitions in {seq_name_list[22]} : \nDistribution : {dict(zip(alphabet_list,count_local_mirror))}, \npercentage: {dict(zip(alphabet_list,percentage_local_mirror))}\n')
print(f'**Counting Local repetitions in {seq_name_list[12]} : \nDistribution : {dict(zip(alphabet_list,count_local_subprog))}, \npercentage: {dict(zip(alphabet_list,percentage_local_subprog))}\n')
print(f'**Counting Local repetitions in {seq_name_list[4]} : \nDistribution : {dict(zip(alphabet_list,count_local_alternate))}, \npercentage: {dict(zip(alphabet_list,percentage_local_alternate))}\n')


```
