```python
data = data_main_complete.copy()
sequences = seq_name_list

# ------------------------------------------
##########  Extract mean DL
# Participants IDs
IDs=[data.iloc[0]["participant_ID"]]
for i in range(len(data)-1):
    if data.iloc[i]["participant_ID"] not in IDs:
        IDs.append(data.iloc[i]["participant_ID"])

# Calculate the mean distance_DL for each sequence per participant
temp_distDL_perParticipant = []

# Gather sequence_expression
sequence_expressions=[dict_expressions[key] for key in sequences]

for name in sequences:
    new_arr = []
    for participant in IDs:
        subset = data[(data["participant_ID"] == participant) & (data["seq_name"] == name)]
        mean_distance_dl = np.nanmean(subset["distance_dl"])  # Use np.nanmean to handle NaN values
        new_arr.append(mean_distance_dl)
            
    temp_distDL_perParticipant.append(new_arr)

# Convert the list of lists into a 2D NumPy array
distDL_perParticipant = np.array(temp_distDL_perParticipant)

# Calculate confidence intervals
CI_meanDL = [confidence_interval95(dist) for dist in distDL_perParticipant]
all_sem = [stats.sem(dist, nan_policy='omit') for dist in distDL_perParticipant]

mean_distDL_perParticipant=[]

for i in range(len(all_sem)):
    # We use np.nanmean because participants from experiment 1 don't have values for sequences tested in experiment 2
    mean_distDL_perParticipant.append(round(np.nanmean(distDL_perParticipant[i]),2))

# 2 - Get complexities
complexities = [complexities_post_fit_exp1[name] for name in seq_name_list]



# ------------------------------------------
##########  Extract mean Subjective complexity
test_sequences_tempTags=[
        'Rep2',
        'CRep2',
        'Rep3',
        'CRep3',
        'Rep4',
       'CRep4',
        'Rep-Nested',
        'Nested-Global-Rep',
        'Nested-Local-Rep',

        'Play4',
        'CPlay4',

         'Sub-1',
        'CSub-1',
        'Sub-2',
        'CSub-2',
       
       'Index',
       'CIndex',
        'Play',
        'CPlay',
       'Insertion',
       'Suppression',

        'Mirror-Rep',
        'CMirror-Rep',
        'Mirror-NoRep',
       'CMirror-NoRep',
]
subj_dataset = pd.read_json("/Users/elyestabbane/Documents/UNICOG/2-Experiments/complexity_estimation/2-Data/online_exp_data/processed/aggregated_data.json")
sequences_tmp = test_sequences_tempTags

# Get unique IDs
IDs_subj = subj_dataset.participant_prolific_id.unique()
all_mean_subj = []
all_sem_subj = []

for seq in sequences_tmp:
    mask = subj_dataset.sequences_temp_tags == seq
    subset_seq = subj_dataset[mask]
    mean_p_subj = []
    
    for idx in IDs_subj:
        mask_participant = subset_seq.participant_prolific_id == idx
        subset_p = subset_seq[mask_participant]
        mean_p_subj.append(np.nanmean(subset_p.participant_response))

    all_mean_subj.append(np.nanmean(mean_p_subj))
    all_sem_subj.append(stats.sem(mean_p_subj, nan_policy='omit'))


# ------------------------------------------
##########  Create dataset
# seq_name / LoT_complexity / mean_DL / sem_DL / mean_subjective / sem_subjective
data_compare_subj = pd.DataFrame({
                'seq_name': test_sequences_tempTags,
                'LoT_complexity': complexities,
                'mean_dl' : mean_distDL_perParticipant,
                'sem_dl' : all_sem,
                'mean_subjective' : all_mean_subj,
                'sem_subjective' : all_sem_subj
})
                
```
