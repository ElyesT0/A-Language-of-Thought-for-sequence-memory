# Computing Chunk Complexity
---

```python
def compute_chunk_complexity(sequence: str) -> float:
    """
    Computes the chunk complexity for a given string.
    A 'chunk' is defined as a maximal run of consecutive repeated characters.

    Complexity = sum_{i=1 to K} [log2(1 + L_i)],
    where K is the number of chunks, and L_i is the length of the i-th chunk.

    Args:
        sequence (str): The input sequence of characters (e.g., "AAABBBAABA").

    Returns:
        float: The computed chunk complexity.
    """
    
    if not sequence:  # Handle empty string
        return 0.0
    
    # Identify chunks
    chunk_sizes = []
    current_chunk_size = 1
    
    for i in range(1, len(sequence)):
        if sequence[i] == sequence[i-1]:
            current_chunk_size += 1
        else:
            chunk_sizes.append(current_chunk_size)
            current_chunk_size = 1
    # Append the last chunk size
    chunk_sizes.append(current_chunk_size)
    
    # Compute complexity: sum_{i=1 to K} log2(1 + L_i)
    complexity_value = sum(math.log2(1 + size) for size in chunk_sizes)
    return complexity_value

    
```

```python


def chunk_complexity_v2(s: str, verbose: bool = False):
    """
    Computes a 'second version' of chunk complexity for string s, where
    we define a chunk as the smallest non-self-repeating building block
    that can tile the entire string by repetition.

    Rules:
      1. We only consider building blocks strictly smaller than the entire 
         string's length OR a building block that is 'self-repeating' 
         (meaning the entire string is repeated copies of a shorter pattern).
      2. If no smaller building block is found that covers the entire string,
         and the string itself is NOT self-repeating, we treat each character 
         as its own chunk.

    Args:
        s (str): The input sequence (e.g., "ABCABCABCABC").
        verbose (bool): If True, prints the identified chunks.

    Returns:
        (float, list[str]): 
            - complexity value
            - identified chunks

    Example:
        s = "122221121112" 
        => no smaller building block covers the string,
        => the entire string is not self-repeating,
        => fallback: each character is its own chunk 
           => complexity = 12 * log2(2) = 12.
    """

    if not s:  # Handle empty string
        if verbose:
            print("Identified chunks: [] (empty string)")
        return 0.0, []

    n = len(s)

    def is_self_repeating(x: str) -> bool:
        """
        Returns True if 'x' can be expressed as a smaller string repeated
        multiple times. Example: "ABCABC" => True, "ABC" => False,
        "AAAA" => True, "ABAB" => True, "A" => False.
        """
        if len(x) == 1:
            return False
        return x in (x + x)[1:-1]

    # 1) Check if the entire string s is a repeated pattern (self-repeating).
    #    If so, we still want to find the minimal block. 
    #    For example, "ABCABCABCABC" => "ABC" repeated 4 times.
    #    We'll search for that minimal block below.
    #    If we can't find a smaller one, we might accept 'ABCABCABCABC' 
    #    if it itself is repeating a smaller string. If not, fallback.

    # 2) Loop through divisors of n from 1 up to n-1 (not including n).
    #    We want to find a strictly smaller building block if it exists.
    for d in range(1, n):
        if n % d == 0:
            candidate = s[:d]          # potential building block
            repeats = n // d
            # Check if s == candidate repeated 'repeats' times
            if candidate * repeats == s:
                # Now check if candidate is self-repeating
                if is_self_repeating(candidate):
                    # candidate is not minimal -> skip it
                    continue
                else:
                    # Found the smallest non-self-repeating building block
                    identified_chunks = [candidate] * repeats
                    complexity = repeats * math.log2(1 + len(candidate))
                    if verbose:
                        print(f"Identified chunks: {identified_chunks}")
                    return complexity, identified_chunks

    # 3) If we didn't return by now, it means:
    #    - No strictly smaller building block tiles the string, or
    #    - They all were self-repeating themselves.
    #
    #    We check if the entire string is self-repeating. If yes,
    #    we accept that (the entire string) as the building block.
    #    Otherwise, fallback to single-character chunks.

    # Entire string as candidate
    candidate = s
    if is_self_repeating(candidate):
        # This means s = smaller substring repeated
        # but we haven't found that substring in the loop above (maybe it's a power?).
        # In that case, s is the building block repeated 1 time, 
        # but s is still self repeating, so let's attempt to find 
        # the *true* smallest substring. One way is to do the same
        # check we did in `is_self_repeating` using (s + s)[1:-1].
        
        # But let's do a direct approach: find the smaller substring 
        # from (s + s)[1:-1].
        # However, if we want to remain consistent, let's just treat the entire 
        # string as 1 chunk if it's "fully self repeating" 
        # -> or we could re-run the logic. 

        # For simplicity, let's treat 's' as 1 chunk if it's self repeating. 
        # But you could refine it further to actually find that smallest substring 
        # from is_self_repeating. 
        identified_chunks = [candidate]  # one big chunk
        complexity = math.log2(1 + n)
        if verbose:
            print(f"Identified chunks: {identified_chunks} (self-repeating entire string)")
        return complexity, identified_chunks
    else:
        # 4) Fallback: each character is its own chunk
        identified_chunks = list(s)
        complexity = len(s) * math.log2(2)  # n * 1
        if verbose:
            print(f"Identified chunks (fallback): {identified_chunks}")
        return complexity, identified_chunks


all_exp=[strv for strv in dict_expressions.values()]
all_names=[name for name in dict_expressions.keys()]
step=1
for k in range(len(all_exp)):
    print(f'{all_names[k]} [{chunk_complexity_v2(all_exp[k])}]')
    if step==2:
        print(f'\n')
        step=0
    step+=1
```

```python
import math

def is_self_repeating(x: str) -> bool:
    """
    Returns True if 'x' can be expressed as a smaller string repeated multiple times.
    Examples:
        - "ABCABC" -> True  (it's "ABC" repeated twice)
        - "AAAA"   -> True  (it's "A" repeated 4 times)
        - "ABAB"   -> True  (it's "AB" repeated twice)
        - "ABC"    -> False
        - "A"      -> False
    """
    if len(x) < 2:
        return False
    # Common trick: x in (x+x)[1:-1] means x is a rotation of itself
    # that suggests repeated pattern. 
    return x in (x + x)[1:-1]

def find_repeated_patterns(s: str):
    """
    Finds all non-self-repeating substrings of 's' that appear 
    at least twice in a NON-overlapping manner.

    Returns:
        A list of (substring, length_of_substring).
        Example: [("123", 3), ("12", 2), ("A", 1), ...]
    """
    n = len(s)
    patterns = set()
    
    # Naive O(n^3) approach: 
    # 1) Enumerate all substrings s[i:j]
    # 2) Count how many non-overlapping times it appears
    # 3) Keep if >= 2 times AND not self-repeating
    for i in range(n):
        for j in range(i+1, n+1):
            substr = s[i:j]
            length_sub = j - i
            # Quick skip for length 1 => can be considered repeated but let's check count below
            # Next, count non-overlapping appearances:
            count = 0
            start_search = 0
            while True:
                # find substring from start_search onward
                found_pos = s.find(substr, start_search)
                if found_pos == -1:
                    break
                count += 1
                # Move start_search by length_sub to ensure non-overlap
                start_search = found_pos + length_sub
            
            if count >= 2:
                # Check if it's self-repeating
                if not is_self_repeating(substr):
                    patterns.add((substr, length_sub))

    # Convert to list and sort by descending length
    # If two patterns have the same length, it doesn't matter which goes first
    patterns_list = sorted(list(patterns), key=lambda x: x[1], reverse=True)
    return patterns_list

def chunk_sequence_local(s: str, patterns_list) -> list:
    """
    Given a sorted list of repeated patterns (descending length),
    parse the string 's' from left to right, greedily matching 
    the longest pattern first. If no pattern matches at a position, 
    fallback to a single-character chunk.

    Returns:
        chunks (list of str)
    """
    chunks = []
    i = 0
    n = len(s)
    
    while i < n:
        matched = False
        # Check each pattern in descending length order
        for (patt, length_patt) in patterns_list:
            if i + length_patt <= n and s[i:i+length_patt] == patt:
                # Found a match -> chunk is 'patt'
                chunks.append(patt)
                i += length_patt
                matched = True
                break
        if not matched:
            # fallback: single char
            chunks.append(s[i])
            i += 1
    return chunks

def compute_complexity(chunks: list) -> float:
    """
    Compute sum_{chunk in chunks} log2(1 + len(chunk)).
    """
    import math
    total = 0.0
    for c in chunks:
        L = len(c)
        total += math.log2(1 + L)
    return total

def chunk_complexity_v2_local(s: str, verbose: bool = False):
    """
    Implements a second version of chunk complexity with a 
    dictionary-based approach to local repeated patterns:
      1) Identify all possible repeated patterns that are not self-repeating.
      2) Sort by descending length.
      3) Greedily parse s from left to right, matching the largest pattern at each step.
      4) Fallback to single-character if no pattern matches.
      5) Compute sum of log2(1 + chunk_length).

    Args:
        s (str): input sequence, e.g. "123412351236"
        verbose (bool): If True, prints the identified chunks.

    Returns:
        (float, list_of_chunks)
    """

    if not s:
        return 0.0, []

    # 1) Find repeated patterns
    patterns_list = find_repeated_patterns(s)

    # 2) Chunk the sequence locally using the dictionary
    identified_chunks = chunk_sequence_local(s, patterns_list)

    # 3) Compute complexity
    complexity_value = compute_complexity(identified_chunks)

    if verbose:
        print(f"All repeated patterns (descending length): {patterns_list}")
        print(f"Identified chunks: {identified_chunks}")
        print(f"Complexity: {complexity_value}")

    return complexity_value, identified_chunks

all_exp=[strv for strv in dict_expressions.values()]
all_names=[name for name in dict_expressions.keys()]
step=1
for k in range(len(all_exp)):
    print(f'{all_names[k]} [{chunk_complexity_v2_local(all_exp[k])}]')
    if step==2:
        print(f'\n')
        step=0
    step+=1
```
