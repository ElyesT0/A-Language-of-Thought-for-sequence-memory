```python
from itertools import takewhile
from scipy import stats
name_list = seq_name_list_exp1_only
pairs = [['control Repetition-3','control Repetition-4'],
        ['Repetition-3','Repetition-4'],
         ['Repetition-2','control Repetition-2']
        ]
data = data_main_complete.copy()

# Compute accurate length per row
data['accurate_length'] = [
    sum(1 for _ in takewhile(lambda pair: pair[0] == pair[1], zip(r, o)))
    for r, o in zip(data.sequences_response, data.seq)
]

# Per-participant mean, then aggregate stats per sequence
per_participant = data.groupby(['participant_ID', 'seq_name'])['accurate_length'].mean()

stats_df = per_participant.groupby('seq_name').agg(
    mean='mean',
    median='median',
    sem='sem'
).round(2)

# Filter to sequences of interest and convert to dict
dict_acc_length = stats_df.loc[name_list].to_dict(orient='index')

# Wilcoxon signed-rank test for ['rep3', 'crep3']
per_participant_df = per_participant.reset_index()


for pair in pairs:
    seq1 = per_participant_df[per_participant_df['seq_name'] == pair[0]]['accurate_length'].values
    seq2 = per_participant_df[per_participant_df['seq_name'] == pair[1]]['accurate_length'].values
    print(f'{pair[0]} : {seq1.mean().round(2)} (SEM : {round(stats.sem(seq1), 2)})')
    print(f'{pair[1]} : {seq2.mean().round(2)} (SEM : {round(stats.sem(seq2), 2)})')
    wilcoxon_stat, p_value = stats.wilcoxon(seq1, seq2)
    print(f"Wilcoxon signed-rank test | {pair[0]} vs {pair[1]}")
    print(f"  Statistic : {wilcoxon_stat:.3f}")
    print(f"  p-value   : {p_value:.3f}")
    print(f"  {'✓ Significant' if p_value < 0.05 else '✗ Not significant'} (α = 0.05)")
    #dict_acc_length
    print(f'\n')
```

```python
data = data_main.copy()
structure_arr = ['Repetition-2','Repetition-3','Repetition-4']
control_arr = ['control Repetition-2','control Repetition-3','control Repetition-4']
structure_mask = "seq_name.isin(@structure_arr)"
control_mask = "seq_name.isin(@control_arr)"

# Compute accurate length per row
data['accurate_length'] = [
    sum(1 for _ in takewhile(lambda x: x[0] == x[1], zip(r, o)))
    for r, o in zip(data.sequences_response, data.seq)
]

data_structure = data.query(structure_mask)
data_control = data.query(control_mask)

# Per-participant means
pp_structure = data_structure.groupby('participant_ID')['accurate_length'].mean()
pp_control = data_control.groupby('participant_ID')['accurate_length'].mean()

# Aggregate stats
stats_structure = pp_structure.agg(['mean', 'median', 'sem']).round(2)
stats_control = pp_control.agg(['mean', 'median', 'sem']).round(2)

print(f'STRUCTURE : mean={stats_structure["mean"]} (SEM: {stats_structure["sem"]})')
print(f'CONTROL   : mean={stats_control["mean"]} (SEM: {stats_control["sem"]})')

wilcoxon_stat, p_value = stats.wilcoxon(pp_structure.values, pp_control.values)
print(f"Wilcoxon signed-rank test | STRUCTURE vs CONTROL")
print(f"  Statistic : {wilcoxon_stat:.3f}")
print(f"  p-value   : {p_value:.3f}")
print(f"  {'✓ Significant' if p_value < 0.05 else '✗ Not significant'} (α = 0.05)")
```
